```python
df = pd.read_csv("Titanic-Dataset.csv")
# Check missing values in the original dataset
print("Before handling missing values\n",df.isnull().sum())
# Fill missing Age values with the median
df["Age"] = df["Age"].fillna(df["Age"].median())
# Fill missing Embarked values with the most common value
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
# Remove Cabin because it has many missing values
df = df.drop(columns=["Cabin"])
# Check missing values again
print("After handling missing values\n",df.isnull().sum())
```

```python
# Count duplicate rows
print("Duplicate rows:", df.duplicated().sum())

# Remove duplicates if any exist
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
# Mean age
print("Average Age:", df["Age"].mean())

# Median age
print("Median Age:", df["Age"].median())

# Most common gender
print("Most Common Gender:", df["Sex"].mode()[0])

# Average ticket fare
print("Average Fare:", df["Fare"].mean())

# Minimum and maximum age
print("Minimum Age:", df["Age"].min())
print("Maximum Age:", df["Age"].max())
```

```python
# Average age by gender
print(df.groupby("Sex")["Age"].mean())

# Average fare by passenger class
print(df.groupby("Pclass")["Fare"].mean())

# Survival rate by gender
print(df.groupby("Sex")["Survived"].mean() * 100)

# Survival rate by passenger class
print(df.groupby("Pclass")["Survived"].mean() * 100)
```
