```python
import pandas as pd
import numpy as np
df = pd.read_csv("Titanic-Dataset.csv")
print("Dataset loaded successfully!")
df = pd.read_csv("Titanic-Dataset.csv")
# Check missing values in the original dataset
print("Before handling missing values\n",df.isnull().sum())
# Fill missing Age values with the median
df["Age"] = df["Age"].fillna(df["Age"].median())
# Fill missing Embarked values with the most common value
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
# Remove Cabin because it has many missing values
df = df.drop(columns=["Cabin"])
# Check missing values again
print("After handling missing values\n",df.isnull().sum())
# Count duplicate rows
print("Duplicate rows:", df.duplicated().sum())
# Remove duplicates if any exist
df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
# Set a simple plotting style
sns.set_style("whitegrid")
```

```python
#Histogram
plt.figure(figsize=(8,5))
sns.histplot(data=df,x="Age",bins=6,color="skyblue")
plt.title("Distribution of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
#Barchart
plt.figure(figsize=(6,4))
sns.countplot(data=df,x="Survived",hue="Survived",palette="Set2",legend=False)
plt.title("Passenger Survival Count")
plt.xlabel("Survival (0 = No,1 = Yes)")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
#scatterplot
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df,x="Age",y="Fare",hue="Survived",palette="coolwarm")
plt.title("Passenger Age vs Ticket Fare")
plt.xlabel("Age")
plt.ylabel("Ticket Fare")
plt.legend(title="Survived")
plt.show()
```

```python
#barchart
plt.figure(figsize=(7,5))
sns.barplot(data=df, x="Sex",y="Survived",hue="Sex",palette="pastel",legend=False)
plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")
plt.show()
```

```python
#correlation Heatmap
# Select numerical columns
numeric_df = df.select_dtypes(include="number")
# Calculate correlations
correlation = numeric_df.corr()
#heatmap
plt.figure(figsize=(10,7))
sns.heatmap(correlation,annot=True,cmap="coolwarm",fmt=".2f")
plt.title("Correlation Heatmap of Titanic Dataset")
plt.show()
```

```python
#custom color palette and plot formation
plt.figure(figsize=(8,5))
sns.countplot(data=df,x="Pclass",hue="Pclass",palette=["#FFD166", "#06D6A0", "#118AB2"],legend=False)
plt.title("Passengers in Each Class",fontsize=15, fontweight="bold")
plt.xlabel("Passenger Class",fontsize=12)
plt.ylabel("Number of Passengers",fontsize=12)
plt.tight_layout()
plt.show()
```
